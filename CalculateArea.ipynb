```python
_dir_name = 'data\\'
file_list = sort_file_list(list_all_files(_dir_name),'png')
meta_data = []

print('\nSTARTED\n')
for _file in file_list:
    print('Processing: ',_file)
    
    
    _file_names = _file
    _file_names = _file_names.replace('.','')
    _file_names = _file_names.replace('_','\\')
    _file_names = _file_names.replace(' ','')
    _file_names = _file_names.upper()
    _file_names = _file_names.split('\\')
    
    _experiment_day = _file_names[1]
    _group = _file_names[2]
    _animal_name = _file_names[-1][:-3]
    _area = get_area_from_image(_file)
    
    _meta_entry = [_experiment_day,_animal_name,_area[0],_area[1],_area[2],_area[3],_group]
    meta_data.append(_meta_entry)
    
    
_this_time = datetime.datetime.now().isoformat(timespec='seconds')
_this_time = str(_this_time)
_this_time = _this_time.replace('-','')
_this_time = _this_time.replace(':','')


save_name = 'ExportAreas'+_this_time+'.csv'
_header = 'Time,Animal,Area1,Area2,Area3,Area4,Group'
np.savetxt(save_name, meta_data, delimiter=",", fmt='%s', header=_header)
print('\nEXPORTED CSV:',save_name)

print('\nData preview\n')
for _entry in meta_data:
    print(_entry)

    
```
